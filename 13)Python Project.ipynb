```python
students = []
```

```python
while True:

    try:

        n = int(input("Enter Number Of Students: "))
        if n <= 0:
            raise ValueError("Number must be positive")
        break
    except ValueError as e:
        print(e)
```

```python
def calculate_score(attendance,quiz,assignment,practice_hours):
    score = (attendance * 0.2 + quiz * 0.3 + assignment * 0.3 + practice_hours * 0.2)
    return round(score,2)
```

```python
for i in range(n):

    print(f"\nStudent {i+1}")

    name = input("Enter Name: ")

    attendance = int(input("Attendance %: "))

    quiz = int(input("Quiz Score: "))

    assignment = int(input("Assignment Score: "))

    practice_hours = int(input("Practice Hours: "))

    student = {

        "name":name,

        "attendance":attendance,

        "quiz":quiz,

        "assignment":assignment,

        "practice_hours":practice_hours
    }

    students.append(student)
```

```python
for student in students:

    score = calculate_score( student["attendance"], student["quiz"], student["assignment"], student["practice_hours"])
    student["score"] = score
```

```python
for student in students:
    if student["score"] >= 80:
        student["status"] = ("Placement Ready")

    elif student["score"] >= 60:
        student["status"] = ("Average")

    else:
        student["status"] = ("At Risk")
```

```python
top_students = sorted(students,key=lambda x:x["score"],reverse=True)
top_students
```

```python
print("BOOTCAMP REPORT")

for student in top_students:

    print(student["name"],"| Score:", student["score"], "|", student["status"])
```

```python
top_student = top_students[0]

average_score = sum(student["score"] for student in students) / len(students)

print(f"Top Performer: {top_student['name']}")
print(f"Average Score: {average_score}")
```

```python
quiz_completed = set()
assignment_completed = set()

for student in students:
    if student["quiz"] >= 40:
        quiz_completed.add(student["name"])

    if student["assignment"] >= 40:
        assignment_completed.add(student["name"])
```

```python
consistent_students = quiz_completed & assignment_completed
print("Consistent Students:")
for student in consistent_students:
    print(student)
```

```python
report = "Bootcamp Performance Report\n\n"

for student in top_students:
    report = report + "Name: " + student["name"] + "\n"
    report = report + "Score: " + str(student["score"]) + "\n"
    report = report + "Status: " + student["status"] + "\n"

report = report + "Top Performer: " + top_student["name"] + "\n"
report = report + "Average Score: " + str(average_score) + "\n"

report = report + "\nConsistent Students:\n"
for student in consistent_students:
    report = report + student + "\n"

file = open("bootcamp_report.txt", "w")
file.write(report)
file.close()

print("Report Saved Successfully")
```
